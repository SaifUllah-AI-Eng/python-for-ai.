## 📝 Explanation

The `print()` function is used to display output on the screen.

Syntax:

```python
print("Text")
```

Everything inside quotation marks is displayed exactly as written.

````python
print("Welcome to Python Programming!")
````

## Example 2

The `print()` function can display any text.

You can change the message inside quotation marks.

````python
print("My name is Ali")
````

````python
print("I am learning Python.")
````

## Multiple Statements

Python executes code from top to bottom.

````python
print("Line 1")
print("Line 2")
print("Line 3")
````

## ⚠️ Common Mistakes

❌ Missing quotation marks

```python
print(Hello)
```

This produces an error because Python thinks `Hello` is a variable.

✔ Correct

```python
print("Hello")
```
